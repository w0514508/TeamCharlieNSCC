## 1. LOAD + CLEAN DATA

```python
import pandas as pd
import numpy as np
from scipy.spatial.distance import cdist
```

```python
import pandas as pd

temp = pd.read_parquet("../../../data/clean/Yashaswi/Final_Clean/temp_2015.parquet")
print(temp.head())
```

```python
precip = pd.read_parquet("../../../data/clean/Yashaswi/Final_Clean/precip_2015_2025.parquet")
print(precip.head())
```

```python
wind = pd.read_parquet("../../../data/clean/Yashaswi/Final_Clean/wind_2015_2025.parquet")
print(wind.head())
```

```python
poll = pd.read_parquet("../../../data/clean/Yashaswi/DimPollutants/Pollutants2015_Parquet/")

poll.rename(columns={'NAPS ID': 'NAPS_ID'}, inplace=True)
print(poll.head())
```

```python
st = pd.read_csv("../../../data/clean/Yashaswi/Final_Clean/dim_station_clean.csv")
print(st.head())
```

```python
naps = pd.read_csv("../../../data/clean/Yashaswi/Final_Clean/Dim_Station_NAPS.csv")
print(naps.head())
```

```python
import pandas as pd
import numpy as np

# --- DATE FORMAT ---
temp['Date'] = pd.to_datetime(temp['Date'])

precip.rename(columns={'Date/Time': 'Date'}, inplace=True)
precip['Date'] = pd.to_datetime(precip['Date'])

wind['Date'] = pd.to_datetime(wind['Date'])
poll['Date'] = pd.to_datetime(poll['Date'])

# --- STANDARDIZE CLIMATE ID ---
temp.rename(columns={'Climate ID': 'Climate_ID'}, inplace=True)
precip.rename(columns={'Climate ID': 'Climate_ID'}, inplace=True)
wind.rename(columns={'Climate ID': 'Climate_ID'}, inplace=True)

```

## ✅ 2. MAP NAPS → CLIMATE ID

```python
climate_st = st[['Climate ID', 'Latitude (y)', 'Longitude (x)']].copy()
climate_st.columns = ['Climate_ID', 'lat_climate', 'lon_climate']

naps_st = naps[['NAPS_ID', 'Latitude', 'Longitude']].copy()
naps_st.columns = ['NAPS_ID', 'lat_naps', 'lon_naps']

dist_matrix = cdist(
    naps_st[['lat_naps','lon_naps']],
    climate_st[['lat_climate','lon_climate']]
)

nearest_idx = dist_matrix.argmin(axis=1)
naps_st['Climate_ID'] = climate_st.iloc[nearest_idx]['Climate_ID'].values
```

## 3. ADD CLIMATE_ID TO POLL

```python
poll = poll.merge(
    naps_st[['NAPS_ID','Climate_ID']],
    on='NAPS_ID',
    how='left'
)
```

## ✅ 4. PIVOT POLL (IMPORTANT ✅)

```python
poll_pivot = poll.pivot_table(
    index=['Date','Climate_ID','NAPS_ID'],
    columns='Pollutant',
    values='C'
).reset_index()
```

## 5. MERGE ALL DATASETS

```python
df = temp.merge(precip, on=['Date','Climate_ID'], how='left') \
         .merge(wind, on=['Date','Climate_ID'], how='left') \
         .merge(poll_pivot, on=['Date','Climate_ID'], how='left')
```

## 6. ADD LOCATION INFO (Province + City)

```python
df = df.merge(
    naps[['NAPS_ID','City','Province/Territory_EN']],
    on='NAPS_ID',
    how='left'
)

df.rename(columns={'Province/Territory_EN': 'Province'}, inplace=True)
```

## 8. FILTER ATLANTIC (CRITICAL STEP)

```python
atlantic = [
    'NEW BRUNSWICK',
    'NEWFOUNDLAND AND LABRADOR',
    'NOVA SCOTIA',
    'PRINCE EDWARD ISLAND'
]

df = df[df['Province'].isin(atlantic)].copy()
```

## ✅ 7. ADD FEATURES (Season, Wind)

```python
# Season
def get_season(m):
    if m in [3,4,5]: return 'Spring'
    elif m in [6,7,8]: return 'Summer'
    elif m in [9,10,11]: return 'Fall'
    else: return 'Winter'

df['month'] = df['Date'].dt.month
df['season'] = df['month'].apply(get_season)

# Wind direction
df['wind_sin'] = np.sin(np.deg2rad(df['Dir of Max Gust (10s deg)']))
df['wind_cos'] = np.cos(np.deg2rad(df['Dir of Max Gust (10s deg)']))
```

```python
proxy_map = {
    'HAPPY VALLEY - GOOSE BAY': 'LABRADOR CITY',
    'PORT AUX CHOIX': 'CORNER BROOK',
    'KEJIMKUJIK': 'KENTVILLE'
}
```

```python
o3_subset = poll_pivot[
    poll_pivot['NAPS_ID'].isin([10601,10801,30502])
][['Date','NAPS_ID','O3']]
```

```python
o3_subset = o3_subset.merge(
    naps[['NAPS_ID','City','Province/Territory_EN']],
    on='NAPS_ID',
    how='left'
)

o3_subset.rename(columns={'Province/Territory_EN':'Province'}, inplace=True)
```

```python
o3_subset['City_proxy'] = o3_subset['City'].replace(proxy_map)
```

```python
weather_cols = [
    'Date','City',
    'Mean Temp (°C)','Total Rain (mm)',
    'Total Snow (cm)','Spd of Max Gust (km/h)',
    'wind_sin','wind_cos'
]

weather_df = df[weather_cols].drop_duplicates()
weather_df.rename(columns={'City':'City_proxy'}, inplace=True)
```

```python
o3_with_weather = pd.merge_asof(
    o3_subset.sort_values('Date'),
    weather_df.sort_values('Date'),
    on='Date',
    by='City_proxy',
    direction='nearest',
    tolerance=pd.Timedelta('2D')
)
```

```python
o3_with_weather.drop(columns=['City_proxy'], inplace=True, errors='ignore')

# add missing columns
for col in df.columns:
    if col not in o3_with_weather.columns:
        o3_with_weather[col] = np.nan

o3_with_weather = o3_with_weather[df.columns]
```

```python
df_final = pd.concat([df, o3_with_weather], ignore_index=True)
```

```python
df_final['Province'].unique()
```

## Identify OUTLIER cities (correct method)

```python
outliers = []

for prov in df_analysis['Province'].unique():
    
    sub = df_analysis[df_analysis['Province'] == prov]
    
    Q1 = sub['PM2.5'].quantile(0.25)
    Q3 = sub['PM2.5'].quantile(0.75)
    IQR = Q3 - Q1
    
    upper = Q3 + 1.5 * IQR
    
    out = sub[sub['PM2.5'] > upper][['City','PM2.5','Province']]
    
    out['Reason'] = 'High PM2.5'
    outliers.append(out)

outliers_df = pd.concat(outliers)
print(outliers_df.head(20))
```

```python
outliers_df['City'].value_counts()
```

## Repeat for key pollutants

```python
pollutants = ['PM2.5','PM10','NO2','O3']

all_outliers = []

for pol in pollutants:
    for prov in df_analysis['Province'].unique():
        
        sub = df_analysis[df_analysis['Province'] == prov]
        
        Q1 = sub[pol].quantile(0.25)
        Q3 = sub[pol].quantile(0.75)
        IQR = Q3 - Q1
        
        upper = Q3 + 1.5 * IQR
        
        out = sub[sub[pol] > upper][['City',pol,'Province']]
        out['Pollutant'] = pol
        
        all_outliers.append(out)

outliers_all = pd.concat(all_outliers)
```

## STEP 1 — City-level correlations

```python
weather_vars = [
    'Mean Temp (°C)',
    'Total Rain (mm)',
    'Total Snow (cm)',
    'Spd of Max Gust (km/h)'
]

pollutants = ['PM2.5','PM10','NO2','CO','O3','SO2']

city_corr_list = []

for city in df_analysis['City'].dropna().unique():
    
    sub = df_analysis[df_analysis['City'] == city]
    
    if len(sub) < 50:   # ignore tiny datasets
        continue
    
    corr = sub[weather_vars + pollutants].corr()
    
    # extract weather → pollutant correlations
    corr_wp = corr.loc[weather_vars, pollutants]
    
    # flatten table
    corr_flat = corr_wp.unstack().reset_index()
    corr_flat.columns = ['Weather','Pollutant','Correlation']
    
    corr_flat['City'] = city
    
    city_corr_list.append(corr_flat)

city_corr_df = pd.concat(city_corr_list)
```

## Find strongest relationships

```python
city_corr_df['Abs_Corr'] = city_corr_df['Correlation'].abs()

top_city_corr = city_corr_df.sort_values('Abs_Corr', ascending=False)

top_city_corr.head(20)
```

```python
def categorize_pm(pm):
    if pm <= 9:
        return 'Good'
    elif pm <= 35.4:
        return 'Moderate'
    elif pm <= 55.4:
        return 'Unhealthy (Sensitive)'
    else:
        return 'Unhealthy'

df['PM_category'] = df['PM2.5'].apply(categorize_pm)
```

```python
def categorize_o3(x):
    if x <= 0.054: return 'Good'
    elif x <= 0.070: return 'Moderate'
    elif x <= 0.085: return 'Unhealthy (Sensitive)'
    elif x <= 0.105: return 'Unhealthy'
    else: return 'Very Unhealthy'

def categorize_pm10(x):
    if x <= 54: return 'Good'
    elif x <= 154: return 'Moderate'
    elif x <= 254: return 'Unhealthy (Sensitive)'
    elif x <= 354: return 'Unhealthy'
    else: return 'Very Unhealthy'

def categorize_co(x):
    if x <= 4.4: return 'Good'
    elif x <= 9.4: return 'Moderate'
    elif x <= 12.4: return 'Unhealthy (Sensitive)'
    elif x <= 15.4: return 'Unhealthy'
    else: return 'Very Unhealthy'

def categorize_no2(x):
    if x <= 53: return 'Good'
    elif x <= 100: return 'Moderate'
    elif x <= 360: return 'Unhealthy (Sensitive)'
    elif x <= 649: return 'Unhealthy'
    else: return 'Very Unhealthy'

def categorize_so2(x):
    if x <= 35: return 'Good'
    elif x <= 75: return 'Moderate'
    elif x <= 185: return 'Unhealthy (Sensitive)'
    elif x <= 304: return 'Unhealthy'
    else: return 'Very Unhealthy'

# Apply
df['O3_cat'] = df['O3'].apply(categorize_o3)
df['PM10_cat'] = df['PM10'].apply(categorize_pm10)
df['CO_cat'] = df['CO'].apply(categorize_co)
df['NO2_cat'] = df['NO2'].apply(categorize_no2)
df['SO2_cat'] = df['SO2'].apply(categorize_so2)
```

```python
moderate_df = df[(df['PM2.5'] > 9) & (df['PM2.5'] <= 35.4)]
```

```python
pollutant_counts = {
    'O3': moderate_df['O3_cat'].value_counts(),
    'PM10': moderate_df['PM10_cat'].value_counts(),
    'CO': moderate_df['CO_cat'].value_counts(),
    'NO2': moderate_df['NO2_cat'].value_counts(),
    'SO2': moderate_df['SO2_cat'].value_counts()
}
```

```python
import pandas as pd

summary = []

for pol, counts in pollutant_counts.items():
    # Count how often pollutant is >= Moderate
    high_count = counts.get('Moderate',0) + counts.get('Unhealthy (Sensitive)',0) + counts.get('Unhealthy',0)
    
    summary.append((pol, high_count))

summary_df = pd.DataFrame(summary, columns=['Pollutant', 'High_Count'])
summary_df = summary_df.sort_values(by='High_Count', ascending=False)

print(summary_df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

top5 = summary_df.head(5)

sns.barplot(
    data=top5,
    x='High_Count',
    y='Pollutant',
    palette='coolwarm'
)

plt.title("Top Pollutants Driving Moderate PM2.5 Conditions")
plt.show()
```

```python
weather_vars = [
    'Mean Temp (°C)',          # ✅ ONLY temp you want
    'Total Rain (mm)',         # ✅ rain
    'Total Snow (cm)',         # ✅ snow
    'Spd of Max Gust (km/h)',  # ✅ wind speed
    'wind_sin',                # ✅ wind direction (component)
    'wind_cos'                 # ✅ wind direction (component)
]
```

```python
pollutants = ['PM2.5', 'PM10', 'NO2', 'CO', 'O3', 'SO2']
```

## ✅ ✅ ✅ STEP 2 — Compute correlation matrix (ONLY weather vs pollutants)

```python
corr = df[weather_vars + pollutants].corr()

# Extract only weather vs pollutant block
wp_corr = corr.loc[weather_vars, pollutants]
```

## ✅ ✅ ✅ STEP 3 — CLEAN visualization (your main plot)
🔥 Heatmap (BEST)

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

sns.heatmap(
    wp_corr,
    annot=True,
    cmap='coolwarm',
    center=0
)

plt.title("Mean Temp, Rain, Snow, Wind vs Pollutants")
plt.show()
```

## STEP 4 — Sort strongest relationships (VERY IMPORTANT)

```python
pairs = wp_corr.unstack().reset_index()
pairs.columns = ['Weather', 'Pollutant', 'Correlation']

strong = pairs.reindex(pairs['Correlation'].abs().sort_values(ascending=False).index)

print(strong.head(10))
```

## Clean bar chart (best summary)

```python
top = strong.head(10)

sns.barplot(
    data=top,
    x='Correlation',
    y=top['Weather'] + " → " + top['Pollutant'],
    palette='coolwarm'
)

plt.axvline(0, color='black')
plt.title("Top Weather–Pollutant Relationships")
plt.show()
```

## STEP 5 — Separate positive vs negative

```python
pairs = wp_corr.unstack().reset_index()
pairs.columns = ['Weather', 'Pollutant', 'Correlation']

positive = pairs[pairs['Correlation'] > 0].sort_values(by='Correlation', ascending=False)
negative = pairs[pairs['Correlation'] < 0].sort_values(by='Correlation')

print("Top Positive:")
print(positive.head(10))

print("\nTop Negative:")
print(negative.head(10))
```

```python
# Positive
sns.barplot(data=positive.head(10),
            x='Correlation',
            y=positive['Weather'] + " → " + positive['Pollutant'])
plt.title("Top Positive Relationships")
plt.show()

# Negative
sns.barplot(data=negative.head(10),
            x='Correlation',
            y=negative['Weather'] + " → " + negative['Pollutant'])
plt.title("Top Negative Relationships")
plt.show()
```

```python
weather_vars = [
    'Mean Temp (°C)',
    'Total Rain (mm)',
    'Total Snow (cm)',
    'Spd of Max Gust (km/h)'
]

pollutants = ['PM2.5','PM10','NO2','CO','O3','SO2']

rows = []
```

```python
for city in df_analysis['City'].dropna().unique():

    city_df = df_analysis[df_analysis['City'] == city]
    province = city_df['Province'].iloc[0]

    for year in city_df['Year'].unique():
        for month in city_df['Month'].unique():

            sub = city_df[
                (city_df['Year'] == year) &
                (city_df['Month'] == month)
            ]

            # Skip if insufficient data
            if len(sub) < 10:
                continue

            corr = sub[weather_vars + pollutants].corr()
            corr_wp = corr.loc[weather_vars, pollutants]

            for w in weather_vars:
                for p in pollutants:
                    rows.append([
                        city,
                        province,
                        year,
                        month,
                        w,
                        p,
                        corr_wp.loc[w, p]
                    ])
```
